Explica la diferencia entre INNER JOIN, LEFT JOIN, RIGHT JOIN y FULL JOIN.

INNER JOIN:(Unión interna)
Devuelve solo las filas que tienen coincidencias en ambas tablas.

Ejemplo
SELECT Clientes.IDCliente, Clientes.Nombre, Pedidos.Producto
FROM Clientes
INNER JOIN Pedidos ON Clientes.IDCliente = Pedidos.IDCliente;

LEFT JOIN (Unión externa izquierda)
Devuelve todas las filas de la tabla izquierda y las coincidencias de la tabla derecha. Si no hay coincidencia, los valores de la tabla derecha aparecen como NULL

Ejemplo: Queremos listar todos los clientes, incluyendo los que no han realizado pedidos:
SELECT Clientes.IDCliente, Clientes.Nombre, Pedidos.Producto
FROM Clientes
LEFT JOIN Pedidos ON Clientes.IDCliente = Pedidos.IDCliente;


RIGHT JOIN (Unión externa derecha)
Devuelve todas las filas de la tabla derecha y las coincidencias de la tabla izquierda. Si no hay coincidencia, los valores de la tabla izquierda aparecen como NULL.

Ejemplo: Queremos listar todos los pedidos, incluso si no hay cliente registrado:
SELECT Clientes.IDCliente, Clientes.Nombre, Pedidos.Producto
FROM Clientes
RIGHT JOIN Pedidos ON Clientes.IDCliente = Pedidos.IDCliente;


FULL JOIN (Unión externa completa)
Devuelve todas las filas de ambas tablas. Si no hay coincidencias, los valores de la otra tabla aparecen como NULL
Ejemplo: Queremos listar todos los clientes y pedidos, incluyendo los clientes sin pedidos y los pedidos sin clientes:
SELECT Clientes.IDCliente, Clientes.Nombre, Pedidos.Producto
FROM Clientes
FULL JOIN Pedidos ON Clientes.IDCliente = Pedidos.IDCliente;

¿Qué hace la cláusula GROUP BY en SQL?
La cláusula GROUP BY se usa para agrupar filas que tienen el mismo valor en una o más columnas y aplicar funciones de agregación como SUM(), COUNT(), AVG(), MAX(), MIN(), etc.

¿Cuándo se usa GROUP BY?
Se usa cuando necesitamos calcular métricas o estadísticas sobre grupos de datos en lugar de registros individuales.
Ejemplo 1: Contar ventas por cliente
Si queremos saber cuántos pedidos ha realizado cada cliente en Sales.SalesOrderHeader de AdventureWorks:

SELECT CustomerID, COUNT(SalesOrderID) AS TotalPedidos
FROM Sales.SalesOrderHeader
GROUP BY CustomerID;

Diferencia entre WHERE y HAVING en SQL
Ambas cláusulas se usan para filtrar datos en SQL, pero se aplican en momentos diferentes del procesamiento de la consulta.

WHERE Antes de agrupar
HAVING  Después de agrupar	

Ejemplo 1: WHERE filtra antes de agrupar
SELECT CustomerID, COUNT(SalesOrderID) AS TotalPedidos
FROM Sales.SalesOrderHeader
WHERE CustomerID > 20000
GROUP BY CustomerID;


Ejemplo 2: HAVING filtra después de agrupar
Si queremos ver los clientes que han hecho más de 5 pedidos, usamos HAVING porque COUNT() es una función de agregación:
SELECT CustomerID, COUNT(SalesOrderID) AS TotalPedidos
FROM Sales.SalesOrderHeader
GROUP BY CustomerID
HAVING COUNT(SalesOrderID) > 5;

 Explicación:
✔ Primero se agrupan los pedidos por CustomerID.
✔ Luego HAVING filtra solo los clientes con más de 5 pedidos.


Ejemplo 3: Combinación de WHERE y HAVING
Si queremos contar los pedidos de clientes con CustomerID > 20000, pero solo mostrar los que tienen más de 5 pedidos:

SELECT CustomerID, COUNT(SalesOrderID) AS TotalPedidos
FROM Sales.SalesOrderHeader
WHERE CustomerID > 20000
GROUP BY CustomerID
HAVING COUNT(SalesOrderID) > 5;


Diferencia clave:
✔ WHERE CustomerID > 20000 filtra clientes antes de agrupar.
✔ HAVING COUNT(SalesOrderID) > 5 filtra después de agrupar.

¿Cuál es la diferencia entre una clave primaria y una clave foránea?
Clave Primaria (PRIMARY KEY)

✔ Identifica de manera única cada fila en una tabla.
✔ No puede contener valores nulos (NULL).
✔ Debe ser única (no se pueden repetir valores).
✔ Cada tabla solo puede tener una clave primaria, pero esta puede estar compuesta por varias

Clave Foránea (FOREIGN KEY)
✔ Es una referencia a una clave primaria en otra tabla.
✔ Establece relaciones entre tablas (integridad referencial).
✔ Puede haber valores repetidos y nulos (dependiendo de la relación).
✔ Garantiza que el valor de la clave foránea exista en la tabla referenciada, evitando datos

Explica la diferencia entre UNION y UNION ALL.

La diferencia entre UNION y UNION ALL radica en cómo manejan los registros duplicados cuando combinan resultados de múltiples consultas.

\\UNION (Elimina Duplicados)
Combina los resultados de dos o más consultas y elimina los duplicados automáticamente.
Realiza una operación de "DISTINCT" internamente, lo que puede afectar el rendimiento en conjuntos de datos grandes.

SELECT CustomerID, FirstName FROM Customers
UNION
SELECT CustomerID, FirstName FROM VIPCustomers;

Si un cliente está en ambas tablas, solo aparecerá una vez en el resultado.

\\UNION ALL (Mantiene Duplicados)
Combina los resultados de dos o más consultas sin eliminar duplicados.
Es más rápido que UNION porque no necesita hacer la verificación de duplicados.

SELECT CustomerID, FirstName FROM Customers
UNION ALL
SELECT CustomerID, FirstName FROM VIPCustomers;

¿Qué es una CTE (Common Table Expression)?
Una CTE (Common Table Expression) es una expresión de tabla común que permite definir una consulta temporal dentro de una consulta SQL. Se usa para mejorar la legibilidad y organización del código SQL, especialmente cuando se trabaja con consultas complejas.

¿Cómo se usa una CTE?
Se declara usando la cláusula WITH, seguida de un nombre para la CTE y una consulta que define su contenido. Luego, esta CTE puede ser utilizada en la consulta principal.

WITH NombreCTE AS (
    -- Consulta que define la CTE
    SELECT columna1, columna2
    FROM TablaOriginal
    WHERE condicion
)
SELECT * FROM NombreCTE;

SELECT Nombre, Correo, Telefono, COUNT(*) AS Duplicados
FROM Clientes
GROUP BY Nombre, Correo, Telefono
HAVING COUNT(*) > 1;

WITH CTE_Duplicados AS (
    SELECT *, ROW_NUMBER() OVER (PARTITION BY Nombre, Correo, Telefono ORDER BY ID) AS RowNum

¿Qué es una VIEW (Vista) en SQL Server?
Una vista (VIEW) en SQL Server es una consulta almacenada que actúa como una tabla virtual. Permite presentar datos de una o varias tablas sin necesidad de almacenar físicamente los resultados.

CREATE VIEW VistaClientesActivos AS
SELECT ID, Nombre, Correo
FROM Clientes
WHERE Activo = 1;

Después, puedes usarla como si fuera una tabla:
SELECT * FROM VistaClientesActivos;

¿Cuándo se recomienda usar VIEW?
Las vistas son útiles en varios escenarios:

1. Para simplificar consultas complejas
Si tienes una consulta con muchos JOIN, filtros y cálculos, puedes crear una vista para reutilizarla fácilmente.

CREATE VIEW VentasClientes AS
SELECT c.Nombre, p.ID AS PedidoID, p.Total
FROM Clientes c
JOIN Pedidos p ON c.ID = p.ClienteID;

Luego, puedes consultarla sin necesidad de escribir la consulta completa:
SELECT * FROM VentasClientes WHERE Total > 100;

2. Para mejorar la seguridad y control de acceso
Las vistas pueden ocultar ciertas columnas a los usuarios sin restringir el acceso a toda la tabla.
CREATE VIEW EmpleadosPublicos AS
SELECT ID, Nombre, Cargo
FROM Empleados;


3. Para abstraer cambios en la estructura de la base de datos
Si cambias la estructura de una tabla, pero los usuarios siguen usando la misma vista, sus consultas no se verán afectadas.

Por ejemplo, si renombraste la columna Correo por Email, en lugar de cambiar todas las consultas en tus aplicaciones, puedes actualizar la vista:

ALTER VIEW VistaClientes AS
SELECT ID, Nombre, Email AS Correo
FROM Clientes;

4. Para mejorar el rendimiento con INDEXED VIEWS (Vistas indexadas)
Las vistas indexadas almacenan los datos físicamente y pueden mejorar la velocidad en consultas complejas.
CREATE VIEW VistaVentasMensuales  
WITH SCHEMABINDING  
AS  
SELECT ClienteID, SUM(Total) AS TotalVentas
FROM dbo.Pedidos
GROUP BY ClienteID;

Luego, agregas un índice:
CREATE UNIQUE CLUSTERED INDEX idx_VistaVentasMensuales ON VistaVentasMensuales(ClienteID);

¿Qué es una WINDOW FUNCTION en SQL Server?
son funciones que permiten realizar cálculos sobre un conjunto de filas relacionadas sin agrupar los datos, es decir, sin perder el detalle de cada fila.

ejemplo
SELECT ClienteID, Fecha, Total,
       SUM(Total) OVER (PARTITION BY ClienteID ORDER BY Fecha) AS VentasAcumuladas
FROM Pedidos;

SUM(Total) OVER (...) calcula la suma acumulada de ventas por cada ClienteID.
PARTITION BY ClienteID reinicia el cálculo por cada cliente.
ORDER BY Fecha suma en orden cronológico

Escribe una consulta que use una window function para calcular el total acumulado de ventas por cliente.
SELECT 
    soh.CustomerID,
    soh.SalesOrderID,
    soh.OrderDate,
    soh.TotalDue,
    SUM(soh.TotalDue) OVER (

        ORDER BY soh.OrderDate
    ) AS TotalAcumulado

 ¿Qué es RANK()?
La función RANK() asigna un número de orden a cada fila dentro de una partición, basado en el valor de una columna (o columnas).
👉 Si hay valores repetidos, se les asigna el mismo ranking, pero se saltan los siguientes números.

¿Qué es DENSE_RANK()?
DENSE_RANK() también asigna un número de orden como RANK(), pero NO se salta números si hay empates.
👉 Es una clasificación "compacta".

¿Qué es una transacción en SQL?
Una transacción es un bloque de instrucciones SQL que se ejecutan como una unidad lógica.
Se asegura de que todo se ejecute correctamente, o nada se ejecute si hay errores.

Reglas de una transacción: ACID
Atomicidad: Todo o nada. Si una parte falla, se cancela todo.
Consistencia: Deja la base en un estado válido.
Aislamiento: Las transacciones no interfieren entre sí.
Durabilidad: Si se confirma, los cambios permanecen.

¿Cómo se usan BEGIN TRANSACTION, COMMIT y ROLLBACK?
BEGIN TRANSACTION: Inicia la transacción.

COMMIT: Confirma los cambios si todo salió bien.

ROLLBACK: Revierte los cambios si hubo un error.

Ejemplo: Insertar datos con seguridad

BEGIN TRANSACTION;

BEGIN TRY
    INSERT INTO Clientes (Nombre, Correo) VALUES ('Juan Pérez', 'juan@correo.com');
    INSERT INTO Pedidos (ClienteID, Total) VALUES (999, 200);  -- supongamos que falla
    
    COMMIT;  -- Solo se ejecuta si TODO fue exitoso
END TRY
BEGIN CATCH
    ROLLBACK;  -- Revierte TODO si ocurre algún error
    PRINT 'Hubo un error. Transacción revertida.';
END CATCH;

 ¿Qué pasa aquí?
Si el INSERT en Pedidos falla (por ejemplo, ClienteID 999 no existe), todo se revierte, incluyendo el primer INSERT en Clientes.
Así evitas que la base de datos quede con datos incompletos o inconsistentes.
