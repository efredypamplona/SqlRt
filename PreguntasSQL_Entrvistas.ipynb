¿Qué es SQL y para qué se utiliza?

es un lenguaje de programación diseñado para gestionar y manipular bases de datos relacionales. Se utiliza para realizar consultas, 
insertar, actualizar y eliminar datos, así como para definir y administrar la estructura de las bases de datos.

Explica la diferencia entre INNER JOIN, LEFT JOIN, RIGHT JOIN y FULL JOIN.

INNER JOIN:(Unión interna)
Devuelve solo las filas que tienen coincidencias en ambas tablas.

Ejemplo
SELECT Clientes.IDCliente, Clientes.Nombre, Pedidos.Producto
FROM Clientes
INNER JOIN Pedidos ON Clientes.IDCliente = Pedidos.IDCliente;

LEFT JOIN (Unión externa izquierda)
Devuelve todas las filas de la tabla izquierda y las coincidencias de la tabla derecha. Si no hay coincidencia, los valores de la tabla derecha aparecen como NULL

Ejemplo: Queremos listar todos los clientes, incluyendo los que no han realizado pedidos:
SELECT Clientes.IDCliente, Clientes.Nombre, Pedidos.Producto
FROM Clientes
LEFT JOIN Pedidos ON Clientes.IDCliente = Pedidos.IDCliente;


RIGHT JOIN (Unión externa derecha)
Devuelve todas las filas de la tabla derecha y las coincidencias de la tabla izquierda. Si no hay coincidencia, los valores de la tabla izquierda aparecen como NULL.

Ejemplo: Queremos listar todos los pedidos, incluso si no hay cliente registrado:
SELECT Clientes.IDCliente, Clientes.Nombre, Pedidos.Producto
FROM Clientes
RIGHT JOIN Pedidos ON Clientes.IDCliente = Pedidos.IDCliente;


FULL JOIN (Unión externa completa)
Devuelve todas las filas de ambas tablas. Si no hay coincidencias, los valores de la otra tabla aparecen como NULL
Ejemplo: Queremos listar todos los clientes y pedidos, incluyendo los clientes sin pedidos y los pedidos sin clientes:
SELECT Clientes.IDCliente, Clientes.Nombre, Pedidos.Producto
FROM Clientes
FULL JOIN Pedidos ON Clientes.IDCliente = Pedidos.IDCliente;

¿Qué hace la cláusula GROUP BY en SQL?
La cláusula GROUP BY se usa para agrupar filas que tienen el mismo valor en una o más columnas y aplicar funciones de agregación como SUM(), COUNT(), AVG(), MAX(), MIN(), etc.

¿Cuándo se usa GROUP BY?
Se usa cuando necesitamos calcular métricas o estadísticas sobre grupos de datos en lugar de registros individuales.
Ejemplo 1: Contar ventas por cliente
Si queremos saber cuántos pedidos ha realizado cada cliente en Sales.SalesOrderHeader de AdventureWorks:

SELECT CustomerID, COUNT(SalesOrderID) AS TotalPedidos
FROM Sales.SalesOrderHeader
GROUP BY CustomerID;

Diferencia entre WHERE y HAVING en SQL
Ambas cláusulas se usan para filtrar datos en SQL, pero se aplican en momentos diferentes del procesamiento de la consulta.

WHERE Antes de agrupar
HAVING  Después de agrupar	

Ejemplo 1: WHERE filtra antes de agrupar
SELECT CustomerID, COUNT(SalesOrderID) AS TotalPedidos
FROM Sales.SalesOrderHeader
WHERE CustomerID > 20000
GROUP BY CustomerID;


Ejemplo 2: HAVING filtra después de agrupar
Si queremos ver los clientes que han hecho más de 5 pedidos, usamos HAVING porque COUNT() es una función de agregación:
SELECT CustomerID, COUNT(SalesOrderID) AS TotalPedidos
FROM Sales.SalesOrderHeader
GROUP BY CustomerID
HAVING COUNT(SalesOrderID) > 5;

 Explicación:
✔ Primero se agrupan los pedidos por CustomerID.
✔ Luego HAVING filtra solo los clientes con más de 5 pedidos.


Ejemplo 3: Combinación de WHERE y HAVING
Si queremos contar los pedidos de clientes con CustomerID > 20000, pero solo mostrar los que tienen más de 5 pedidos:

SELECT CustomerID, COUNT(SalesOrderID) AS TotalPedidos
FROM Sales.SalesOrderHeader
WHERE CustomerID > 20000
GROUP BY CustomerID
HAVING COUNT(SalesOrderID) > 5;


Diferencia clave:
✔ WHERE CustomerID > 20000 filtra clientes antes de agrupar.
✔ HAVING COUNT(SalesOrderID) > 5 filtra después de agrupar.

¿Qué es un índice en SQL y cuáles son sus tipos?
Un índice en SQL es una estructura que acelera la búsqueda de datos en una tabla, similar a un índice en un libro. Permite acceder a las filas de manera más rápida sin necesidad de recorrer toda la tabla.
